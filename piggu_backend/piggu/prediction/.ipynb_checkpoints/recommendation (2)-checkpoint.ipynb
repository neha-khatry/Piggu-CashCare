```python
dataset.rename(columns={'Rent Income ':'Rent_Income'}, inplace= True)
```

#Handling Null values

```python
from sklearn.impute import SimpleImputer
income_columns = ['Salary', 'Rent_Income', 'Investment', 'Lottery', 'Refunds', 'Rewards', 'Cashbacks', 'Interest', 'Others']
expense_columns = ['Food', 'Rent Expense', 'Fuel', 'Electricity', 'Water', 'Education', 'Grocery', 'Health', 'Clothing',
                   'Accessories', 'Tax', 'Travel', 'Office', 'Telephone', 'Fitness', 'Transportation', 'Pet',
                   'Gift', 'Movie', 'Furniture', 'Decoration', 'Insurance', 'Electronics', 'Sports', 'Beverage',
                   'Household']

imputer = SimpleImputer(strategy='constant', fill_value=0)
dataset[income_columns + expense_columns] = imputer.fit_transform(dataset[income_columns + expense_columns])
```

```python
dataset.head(10)
```

#calculating total income, expense and savings

```python
dataset.columns
```

```python
dataset['Total_Income'] = dataset[['Salary', 'Rent_Income', 'Investment', 'Lottery',
       'Refunds', 'Rewards', 'Cashbacks', 'Interest', 'Others']].sum(axis=1)

dataset['Total_Expenses'] = dataset[['Food', 'Rent Expense',
       'Fuel', 'Electricity', 'Water', 'Education', 'Grocery', 'Health',
       'Clothing', 'Accessories', 'Tax', 'Travel', 'Office', 'Telephone',
       'Fitness', 'Transportation', 'Pet', 'Gift', 'Movie', 'Furniture',
       'Decoration', 'Insurance', 'Electronics', 'Sports', 'Beverage',
       'Household']].sum(axis=1)

dataset['Savings'] = dataset['Total_Income']- dataset['Total_Expenses']
```

#Normalize the data

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
normalized_data = scaler.fit_transform(dataset[income_columns + expense_columns + ['Total_Income'] + ['Total_Expenses'] + ['Savings']])
```

```python
dataframe = pd.DataFrame(normalized_data, columns=dataset.columns[2:])
```

```python
dataframe.head(50)
```

```python
min_value = dataframe['Total_Expenses'].min()
max_value = dataframe['Total_Expenses'].max()
value_range = max_value - min_value

print(f"Minimum Value: {min_value}")
print(f"Maximum Value: {max_value}")
print(f"Range: {value_range}")
```

```python
plt.figure(figsize=(15,10))
sns.scatterplot(x= dataset['Total_Income'] , y=dataset['Rent_Income'], data=dataset )
```

```python
from sklearn.model_selection import train_test_split

x=dataset[['Total_Income']]
y= dataset['Rent_Income']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(max_depth=5, random_state=42)
model.fit(x_train, y_train)
```

```python
model.score(x_train, y_train)
```

```python
#random forest regressor
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=200, random_state=42)
model.fit(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
model.score(x_train, y_train)
```

```python
#svr
from sklearn.svm import SVR

model = SVR(kernel='rbf')  # Use RBF kernel for non-linear relationships
model.fit(x_train, y_train)
```

```python
model.score(x_train, y_train)
```

```python
from sklearn.neighbors import KNeighborsRegressor

model = KNeighborsRegressor(n_neighbors=5)
model.fit(x_train, y_train)
```

```python
model.score(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
from xgboost import XGBRegressor

model = XGBRegressor()
model.fit(x_train, y_train)
```

```python
model.score(x_train, y_train)
```

#Savings Prediction

```python
import seaborn as sns
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
input = dataset[['Total_Income']]
output= dataset['Savings']

X_train, X_test, y_train, y_test = train_test_split(input, output, test_size=0.2, random_state=42)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(max_depth=12)
regressor.fit(X_train, y_train)
```

```python
regressor.score(X_train, y_train)

```

```python
regressor.score(X_test, y_test)
```

```python
pr = regressor.predict([[51000]])
print(pr)
```

```python
import joblib
model_path = 'saving_model.pkl'
joblib.dump(regressor, model_path)
print(f'Model saved at {model_path}')
```
